# Resolver o quebra-cabeças usando Buscas

```python
from collections import deque
```

## Busca em largura

```python
tabuleiro_teste = (1,2,3,
                   4,5,6,
                   0,7,8)

type(tabuleiro_teste)
```

```python
def mover_esquerda(tabuleiro):
    index = tabuleiro.index(0)
    novo_estado = list(tabuleiro)

    # se for coluna 0, nao pode mover para a esquerda
    if (index % 3) != 0:
        index_trocar = index - 1
        valor_trocado = tabuleiro[index_trocar] # valor a ser trocado
        
        novo_estado[index] = valor_trocado
        novo_estado[index_trocar] = 0
        return novo_estado
    else:
        return None


def mover_direita(tabuleiro):
    index = tabuleiro.index(0)
    novo_estado = list(tabuleiro)

    if (index % 3) != 2:
        index_trocar = index + 1
        valor_trocado = tabuleiro[index_trocar] # valor a ser trocado
        
        novo_estado[index] = valor_trocado
        novo_estado[index_trocar] = 0
        return novo_estado
    else:
        return None



def mover_cima(tabuleiro):
    index = tabuleiro.index(0)
    novo_estado = list(tabuleiro)

    if index >= 3:
        index_trocar = index - 3
        valor_trocado = tabuleiro[index_trocar] # valor a ser trocado
        
        novo_estado[index] = valor_trocado
        novo_estado[index_trocar] = 0
        return novo_estado
    else:
        return None


def mover_baixo(tabuleiro):
    index = tabuleiro.index(0)
    novo_estado = list(tabuleiro)

    if index <= 5:
        index_trocar = index + 3
        valor_trocado = tabuleiro[index_trocar] # valor a ser trocado
        
        novo_estado[index] = valor_trocado
        novo_estado[index_trocar] = 0
        return novo_estado
    else:
        return None
    
```

BFS INICIAL, SEM MODIFICACOES PARA O TABULEIRO

```python
def breadth_first_search (grafo,nodo_inicial):
    fila = deque([nodo_inicial]) # para nodos, que foram descobertos mas nao explorados
    visitados = set() # nodos explorados
    ordem_visita = list ()

    while fila:
        no_atual = fila.popleft()
        visitados.add(no_atual)
        ordem_visita.append(no_atual)

        for vizinho in grafo[no_atual]:
            if vizinho not in visitados:
                fila.append(vizinho)
                visitados.add(vizinho)

    return ordem_visita
```

## Busca em Profundidade

```python
def depth_first_search (grafo,nodo_inicial):
    fila = deque([nodo_inicial]) # para nodos, que foram descobertos mas nao explorados
    visitados = set() # nodos explorados
    ordem_visita = list () # para retornar

    while fila:
        no_atual = fila.pop()
        visitados.add(no_atual)
        ordem_visita.append(no_atual)

        for vizinho in grafo[no_atual]:
            if vizinho not in visitados:
                fila.append(vizinho)
                visitados.add(vizinho)
                

    return ordem_visita
```
