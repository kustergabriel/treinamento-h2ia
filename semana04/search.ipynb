BFS

```python
from collections import deque
```

```python
grafo = {
    'A': ['B', 'C'],
    'B': ['A', 'D', 'E'],
    'C': ['A', 'F'],
    'F': ['C'],
    'D': ['B'],
    'E': ['B']
}
```

BFS utiliza uma fila, first in first out

```python
def breadth_first_search (grafo,nodo_inicial):
    fila = deque([nodo_inicial]) # para nodos, que foram descobertos mas nao explorados
    visitados = set() # nodos explorados
    ordem_visita = list ()

    while fila:
        no_atual = fila.popleft()
        visitados.add(no_atual)
        ordem_visita.append(no_atual)

        for vizinho in grafo[no_atual]:
            if vizinho not in visitados:
                fila.append(vizinho)
                visitados.add(vizinho)
                

    return ordem_visita
```

```python
breadth_first_search(grafo, 'A')
```

DFS

```python
def depth_first_search (grafo,nodo_inicial):
    fila = deque([nodo_inicial]) # para nodos, que foram descobertos mas nao explorados
    visitados = set() # nodos explorados
    ordem_visita = list () # para retornar

    while fila:
        no_atual = fila.pop()
        visitados.add(no_atual)
        ordem_visita.append(no_atual)

        for vizinho in grafo[no_atual]:
            if vizinho not in visitados:
                fila.append(vizinho)
                visitados.add(vizinho)
                

    return ordem_visita
```

```python
depth_first_search(grafo,'A')
```
